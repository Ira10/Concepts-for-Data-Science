| Concept | Where it appeared |
|---|---|
| **LLM generates text** | `response.choices[0].message.content` |
| **Tokens** | Every word in your message was tokenized before being sent |
| **Temperature** | `temperature=0.0` vs `temperature=2.0` |
| **Context window** | `messages=[...]` — everything you send counts against it |
| **Prompts and roles** | `{"role": "user", "content": "..."}` |

```python
import os
from openai import OpenAI

# Create a client using your KodeKey credentials
client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY"),
    base_url=os.getenv("OPENAI_API_BASE")
)

# Send a message to the LLM
response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {"role": "user", "content": "Explain what an API is in one sentence"}
    ]
)

# Print the response
print(response.choices[0].message.content)
```
